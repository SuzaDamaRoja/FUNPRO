#FUNCIONES
Estructura de programación que permite "cortar" un código grande en partes pequeñas y funcionales.

sintaxis de una función

```
def nombre_de_la_función(d,d1,d2,d3,...,dn):
  operacion1
  operación2
  funcion1
  estructura de control
  estructura de selección
  return variable
```
donde:
* d,d1,d2: son los parámetros de la función
* Por dentro pueden ir todas las operaciones
* return vuelve una variable local -->global

Existen 4 tipos de funciones
1. Sin parámetro, sin retorno.
```
def funcion1():
  operacion1
  operacion2
```
2. Sin parámetro, con retorno.
```
def funcion2():
  operacion1
  operacion2
  return dato
```
3. Con parámetro, sin retorno.
```
def funcion3(color):
  operacion1
  operacion2
```
4. Con parámetro, con retorno.
```
def funcion4(n1,n2,n3,n4):
  operacion1
  operacion2
  return promedio
```

````python
def promedio():
  notas=[]
  n=int(input("Ingrese la cantidad de notas:"))
  for i in range(n):
    nota=float(input(f"Ingrese la nota {i+1}:"))
    notas.append(nota)
  prom=sum(notas)/n
  return notas,prom
````

````python
def max_min(notas):
  print(f"El estudiante obtuvo como mayor nota{max(notas)} y como peor {min(notas)}")
````

````python
def contar(notas):
  c=0
  for i in notas:
    if i>=3: c=c+1
  print(f"De forma adicional, el estudiante obtuvo {c} notas mayor o igual a 3 de {len(notas)}")
````
